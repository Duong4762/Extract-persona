````python
import gzip
import hashlib
import json
import math
import os
import random
import re
import sqlite3
import time
from collections import Counter, defaultdict
from datetime import datetime, timezone
from pathlib import Path
from typing import Any, Iterable, Iterator

import requests
from tqdm.auto import tqdm

WORK = Path("amazon_persona_fresh")
WORK.mkdir(parents=True, exist_ok=True)
DB_PATH = WORK / "amazon.sqlite"

CATEGORIES = [
    "Books", "Kindle_Store", "Movies_and_TV", "Electronics",
    "Office_Products", "Home_and_Kitchen", "Clothing_Shoes_and_Jewelry",
]
START_YEAR, END_YEAR = 2018, 2023

MAX_ROWS_PER_CATEGORY = 50_000
TOP_K = 100
TRAIN_FRACTION = 0.8

MIN_REVIEWS = 30
MIN_CATEGORIES = 2
MIN_HISTORY_DAYS = 365
MIN_TEXT_CHARS = 5_000
MIN_VERIFIED_SHARE = 0.70

REVIEW_URL = (
    "https://mcauleylab.ucsd.edu/public_datasets/data/amazon_2023/"
    "raw/review_categories/{category}.jsonl.gz"
)
META_URL = (
    "https://mcauleylab.ucsd.edu/public_datasets/data/amazon_2023/"
    "raw/meta_categories/meta_{category}.jsonl.gz"
)

print("Work directory:", WORK)
````

````python
SAMPLE_CATEGORY = CATEGORIES[0]


def read_first_remote_row(url: str) -> dict[str, Any]:
    with requests.get(url, stream=True, timeout=120) as response:
        response.raise_for_status()
        response.raw.decode_content = False
        with gzip.GzipFile(fileobj=response.raw) as gz:
            return json.loads(next(line for line in gz if line.strip()))


review_sample = read_first_remote_row(
    REVIEW_URL.format(category=SAMPLE_CATEGORY)
)
meta_sample = read_first_remote_row(
    META_URL.format(category=SAMPLE_CATEGORY)
)

print(f"===== RAW SAMPLE: {SAMPLE_CATEGORY} =====")
print("\nREVIEW:")
print(json.dumps(review_sample, ensure_ascii=False, indent=2))
print("\nMETA:")
print(json.dumps(meta_sample, ensure_ascii=False, indent=2))
````

````python
def stream_gzip_jsonl(url: str, timeout: int = 120) -> Iterator[dict[str, Any]]:
    """Stream a remote .jsonl.gz without saving the compressed file."""
    with requests.get(url, stream=True, timeout=timeout) as response:
        response.raise_for_status()
        response.raw.decode_content = False
        with gzip.GzipFile(fileobj=response.raw) as gz:
            for raw_line in gz:
                if raw_line.strip():
                    yield json.loads(raw_line)


def timestamp_ms(value: Any) -> int | None:
    try:
        number = int(value)
    except (TypeError, ValueError):
        return None
    return number * 1000 if number < 10_000_000_000 else number


def year_of(ts: int) -> int:
    return datetime.fromtimestamp(ts / 1000, tz=timezone.utc).year


def review_id(row: dict[str, Any], category: str) -> str:
    supplied = row.get("review_id")
    if supplied:
        return str(supplied)
    identity = "|".join(str(row.get(k) or "") for k in
                        ("user_id", "parent_asin", "asin", "timestamp", "rating", "text"))
    return hashlib.sha1(f"{category}|{identity}".encode()).hexdigest()


def connect() -> sqlite3.Connection:
    db = sqlite3.connect(DB_PATH)
    db.execute("PRAGMA journal_mode=WAL")
    db.execute("PRAGMA synchronous=NORMAL")
    return db


def create_tables(db: sqlite3.Connection) -> None:
    db.executescript("""
    CREATE TABLE IF NOT EXISTS reviews (
      review_id TEXT PRIMARY KEY, user_id TEXT NOT NULL, category TEXT NOT NULL,
      parent_asin TEXT, asin TEXT, ts INTEGER NOT NULL, rating REAL,
      title TEXT, text TEXT, verified INTEGER, helpful INTEGER,
      product_title TEXT, product_main_category TEXT, product_category_path TEXT
    );
    CREATE INDEX IF NOT EXISTS idx_reviews_user_ts ON reviews(user_id, ts);
    CREATE INDEX IF NOT EXISTS idx_reviews_parent ON reviews(category, parent_asin);
    CREATE TABLE IF NOT EXISTS selected_users (
      rank INTEGER PRIMARY KEY, user_id TEXT UNIQUE, score REAL,
      review_count INTEGER, category_count INTEGER, text_chars INTEGER,
      history_days REAL, verified_share REAL
    );
    """)
    db.commit()


db = connect()
create_tables(db)
print("Database ready:", DB_PATH)
````

````python
insert_review = """
INSERT OR IGNORE INTO reviews
(review_id,user_id,category,parent_asin,asin,ts,rating,title,text,verified,helpful)
VALUES (?,?,?,?,?,?,?,?,?,?,?)
"""

for category in CATEGORIES:
    buffered, scanned, kept = [], 0, 0
    print(f"Streaming {category}...")
    for row in stream_gzip_jsonl(REVIEW_URL.format(category=category)):
        scanned += 1
        if MAX_ROWS_PER_CATEGORY and scanned > MAX_ROWS_PER_CATEGORY:
            break
        user_id = str(row.get("user_id") or "").strip()
        ts = timestamp_ms(row.get("timestamp", row.get("sort_timestamp")))
        if not user_id or ts is None or not START_YEAR <= year_of(ts) <= END_YEAR:
            continue
        text = str(row.get("text") or "").strip()
        title = str(row.get("title") or "").strip()
        buffered.append((
            review_id(row, category), user_id, category,
            str(row.get("parent_asin") or ""), str(row.get("asin") or ""), ts,
            row.get("rating"), title, text,
            int(bool(row.get("verified_purchase"))),
            int(row.get("helpful_vote", row.get("helpful_votes", 0)) or 0),
        ))
        kept += 1
        if len(buffered) >= 5_000:
            db.executemany(insert_review, buffered); db.commit(); buffered.clear()
    if buffered:
        db.executemany(insert_review, buffered); db.commit()
    print(f"{category}: scanned={scanned:,}, eligible-window rows={kept:,}")

print("Stored reviews:", db.execute("SELECT COUNT(*) FROM reviews").fetchone()[0])
````

````python
rows = db.execute("""
SELECT user_id, COUNT(*), COUNT(DISTINCT category),
       SUM(CASE WHEN length(trim(text))>0 THEN 1 ELSE 0 END),
       SUM(length(COALESCE(text,''))),
       (MAX(ts)-MIN(ts))/86400000.0,
       AVG(verified)
FROM reviews GROUP BY user_id
HAVING COUNT(*) >= ? AND COUNT(DISTINCT category) >= ?
   AND (MAX(ts)-MIN(ts))/86400000.0 >= ?
   AND SUM(length(COALESCE(text,''))) >= ? AND AVG(verified) >= ?
""", (MIN_REVIEWS, MIN_CATEGORIES, MIN_HISTORY_DAYS,
      MIN_TEXT_CHARS, MIN_VERIFIED_SHARE)).fetchall()

def percentile_ranks(values: list[float]) -> list[float]:
    order = sorted(range(len(values)), key=values.__getitem__)
    result = [0.0] * len(values)
    i = 0
    while i < len(order):
        j = i + 1
        while j < len(order) and values[order[j]] == values[order[i]]: j += 1
        pct = ((i + 1 + j) / 2) / len(order)
        for k in order[i:j]: result[k] = pct
        i = j
    return result

metrics = [[float(row[col] or 0) for row in rows] for col in range(1, 7)]
# columns: review_count, category_count, text_reviews, text_chars, days, verified
weights = [0.05, 0.20, 0.20, 0.35, 0.15, 0.05]
ranks = [percentile_ranks(metric) for metric in metrics]
scored = []
for i, row in enumerate(rows):
    score = sum(weights[j] * ranks[j][i] for j in range(6))
    scored.append((score, row))
scored.sort(key=lambda item: (-item[0], -item[1][4], item[1][0]))

db.execute("DELETE FROM selected_users")
for rank_no, (score, row) in enumerate(scored[:TOP_K], 1):
    user_id, reviews, categories, text_reviews, text_chars, days, verified = row
    db.execute("INSERT INTO selected_users VALUES (?,?,?,?,?,?,?,?)",
               (rank_no, user_id, score, reviews, categories, text_chars, days, verified))
db.commit()
print(f"Eligible={len(rows):,}; selected={min(TOP_K,len(scored)):,}")
````

````python
import json
import time
import threading

from concurrent.futures import ThreadPoolExecutor, as_completed
from typing import Any


# ============================================================
# CẤU HÌNH
# ============================================================

MAX_METADATA_WORKERS = 4

# In log sau mỗi số dòng này
METADATA_LOG_EVERY_ROWS = 50_000

# Hoặc in log nếu đã qua số giây này
METADATA_LOG_EVERY_SECONDS = 30

# Số bản ghi cập nhật SQLite trong mỗi batch
SQLITE_UPDATE_BATCH_SIZE = 2_000


# Khóa để log từ nhiều thread không bị dính vào nhau
print_lock = threading.Lock()


def log(message: str) -> None:
    """In log an toàn khi chạy đa luồng."""
    with print_lock:
        print(message, flush=True)


# ============================================================
# LẤY DANH SÁCH SẢN PHẨM CẦN TÌM
# ============================================================

def wanted_products(category: str) -> set[str]:
    """
    Lấy danh sách parent_asin duy nhất thuộc một category
    đã được review bởi selected users.
    """
    return {
        str(row[0]).strip()
        for row in db.execute(
            """
            SELECT DISTINCT r.parent_asin
            FROM reviews r
            JOIN selected_users s
                ON s.user_id = r.user_id
            WHERE r.category = ?
              AND r.parent_asin IS NOT NULL
              AND length(r.parent_asin) > 0
            """,
            (category,),
        )
        if row[0]
    }


# ============================================================
# XỬ LÝ METADATA CỦA MỘT CATEGORY
# ============================================================

def fetch_category_metadata(
    category: str,
    wanted: set[str],
) -> tuple[str, dict[str, dict[str, Any]], set[str], int]:
    """
    Đọc file metadata của một category.

    Chỉ giữ metadata của những parent_asin nằm trong wanted.
    Hàm này không ghi SQLite vì được chạy trong worker thread.
    """

    remaining = set(wanted)
    found: dict[str, dict[str, Any]] = {}

    scanned = 0
    start_time = time.time()
    last_log_time = start_time
    first_row_received = False

    url = META_URL.format(category=category)

    log(
        f"[{category}] BẮT ĐẦU — "
        f"cần tìm {len(wanted):,} sản phẩm"
    )

    log(
        f"[{category}] Đang kết nối và mở file metadata..."
    )

    try:
        for meta in stream_gzip_jsonl(url):
            scanned += 1
            current_time = time.time()

            if not first_row_received:
                first_row_received = True

                log(
                    f"[{category}] Đã nhận dòng metadata đầu tiên"
                )

            parent_asin = str(
                meta.get("parent_asin") or ""
            ).strip()

            # Không phải sản phẩm cần tìm
            if parent_asin in remaining:
                found[parent_asin] = {
                    "product_title": str(
                        meta.get("title") or ""
                    )[:500],

                    "product_main_category": str(
                        meta.get("main_category") or ""
                    )[:200],

                    "product_category_path": (
                        meta.get("categories") or []
                    ),
                }

                remaining.remove(parent_asin)

                log(
                    f"[{category}] TÌM THẤY "
                    f"{len(found):,}/{len(wanted):,} — "
                    f"parent_asin={parent_asin}"
                )

            # In tiến độ theo số dòng hoặc thời gian
            should_log_by_rows = (
                scanned % METADATA_LOG_EVERY_ROWS == 0
            )

            should_log_by_time = (
                current_time - last_log_time
                >= METADATA_LOG_EVERY_SECONDS
            )

            if should_log_by_rows or should_log_by_time:
                elapsed = current_time - start_time

                speed = (
                    scanned / elapsed
                    if elapsed > 0
                    else 0
                )

                progress = (
                    len(found) / len(wanted) * 100
                    if wanted
                    else 100
                )

                log(
                    f"[{category}] ĐANG CHẠY — "
                    f"đã quét {scanned:,} dòng | "
                    f"tìm thấy {len(found):,}/{len(wanted):,} "
                    f"({progress:.1f}%) | "
                    f"còn thiếu {len(remaining):,} | "
                    f"{speed:,.0f} dòng/giây | "
                    f"{elapsed / 60:.1f} phút"
                )

                last_log_time = current_time

            # Đã tìm đủ thì không cần đọc phần còn lại
            if not remaining:
                elapsed = time.time() - start_time

                log(
                    f"[{category}] ĐÃ TÌM ĐỦ — "
                    f"dừng sớm tại dòng {scanned:,} | "
                    f"{elapsed / 60:.1f} phút"
                )

                break

    except Exception as error:
        elapsed = time.time() - start_time

        log(
            f"[{category}] LỖI KHI ĐỌC METADATA — "
            f"{error!r} | "
            f"đã quét {scanned:,} dòng | "
            f"{elapsed / 60:.1f} phút"
        )

        # Ném lỗi ra ngoài để luồng chính nhận biết
        raise

    elapsed = time.time() - start_time

    log(
        f"[{category}] HOÀN TẤT — "
        f"tìm thấy {len(found):,}/{len(wanted):,} | "
        f"thiếu {len(remaining):,} | "
        f"đã quét {scanned:,} dòng | "
        f"{elapsed / 60:.1f} phút"
    )

    return category, found, remaining, scanned


# ============================================================
# LẤY DANH SÁCH SẢN PHẨM THEO CATEGORY
# ============================================================

log("[MAIN] Đang lấy danh sách sản phẩm cần tìm từ SQLite...")

wanted_by_category: dict[str, set[str]] = {}

for category in CATEGORIES:
    wanted = wanted_products(category)

    if wanted:
        wanted_by_category[category] = wanted

        log(
            f"[MAIN] {category}: "
            f"cần tìm {len(wanted):,} sản phẩm"
        )
    else:
        log(
            f"[MAIN] {category}: "
            f"không có sản phẩm cần tìm — bỏ qua"
        )

total_wanted = sum(
    len(wanted)
    for wanted in wanted_by_category.values()
)

log(
    f"[MAIN] Tổng cộng {len(wanted_by_category):,} category, "
    f"{total_wanted:,} sản phẩm cần tìm"
)


# ============================================================
# CHẠY ĐA LUỒNG
# ============================================================

metadata_results: list[
    tuple[str, dict[str, dict[str, Any]], set[str], int]
] = []

failed_categories: list[tuple[str, str]] = []

overall_start_time = time.time()

log(
    f"[MAIN] Bắt đầu xử lý metadata bằng "
    f"{MAX_METADATA_WORKERS} worker..."
)

with ThreadPoolExecutor(
    max_workers=MAX_METADATA_WORKERS,
    thread_name_prefix="metadata-worker",
) as executor:

    futures = {
        executor.submit(
            fetch_category_metadata,
            category,
            wanted,
        ): category
        for category, wanted in wanted_by_category.items()
    }

    log(
        f"[MAIN] Đã gửi {len(futures):,} tác vụ vào ThreadPool"
    )

    completed_count = 0

    for future in as_completed(futures):
        category = futures[future]
        completed_count += 1

        try:
            result = future.result()
            metadata_results.append(result)

            _, found, remaining, scanned = result

            log(
                f"[MAIN] HOÀN THÀNH "
                f"{completed_count:,}/{len(futures):,} category — "
                f"{category} | "
                f"tìm thấy {len(found):,} | "
                f"thiếu {len(remaining):,} | "
                f"quét {scanned:,} dòng"
            )

        except Exception as error:
            failed_categories.append(
                (category, repr(error))
            )

            log(
                f"[MAIN] THẤT BẠI "
                f"{completed_count:,}/{len(futures):,} — "
                f"{category}: {error!r}"
            )


metadata_elapsed = time.time() - overall_start_time

log(
    f"[MAIN] Đã xử lý xong metadata — "
    f"thành công {len(metadata_results):,}/"
    f"{len(futures):,} category | "
    f"thất bại {len(failed_categories):,} | "
    f"{metadata_elapsed / 60:.1f} phút"
)


# ============================================================
# GHI KẾT QUẢ VÀO SQLITE
# Chỉ luồng chính thực hiện phần này
# ============================================================

update_product_sql = """
UPDATE reviews
SET product_title = ?,
    product_main_category = ?,
    product_category_path = ?
WHERE category = ?
  AND parent_asin = ?
"""


def update_sqlite_in_batches(
    category: str,
    found: dict[str, dict[str, Any]],
) -> int:
    """Ghi metadata vào SQLite theo từng batch."""

    updates = [
        (
            meta["product_title"],
            meta["product_main_category"],
            json.dumps(
                meta["product_category_path"],
                ensure_ascii=False,
            )[:2000],
            category,
            parent_asin,
        )
        for parent_asin, meta in found.items()
    ]

    if not updates:
        log(
            f"[{category}] Không có metadata để ghi vào SQLite"
        )
        return 0

    total = len(updates)
    written = 0

    log(
        f"[{category}] Bắt đầu ghi "
        f"{total:,} sản phẩm vào SQLite..."
    )

    try:
        for start in range(
            0,
            total,
            SQLITE_UPDATE_BATCH_SIZE,
        ):
            batch = updates[
                start:start + SQLITE_UPDATE_BATCH_SIZE
            ]

            db.executemany(
                update_product_sql,
                batch,
            )
            db.commit()

            written += len(batch)

            log(
                f"[{category}] Đã ghi "
                f"{written:,}/{total:,} sản phẩm vào SQLite"
            )

    except Exception:
        db.rollback()

        log(
            f"[{category}] LỖI KHI GHI SQLITE — "
            f"đã rollback batch hiện tại"
        )

        raise

    return written


log(
    f"[MAIN] Bắt đầu ghi kết quả của "
    f"{len(metadata_results):,} category vào SQLite..."
)

total_found = 0
total_missing = 0
total_written = 0
total_scanned = 0

for category, found, remaining, scanned in metadata_results:
    written = update_sqlite_in_batches(
        category=category,
        found=found,
    )

    total_found += len(found)
    total_missing += len(remaining)
    total_written += written
    total_scanned += scanned


# ============================================================
# TỔNG KẾT
# ============================================================

overall_elapsed = time.time() - overall_start_time

log("")
log("================ KẾT QUẢ ================")
log(
    f"Category thành công : {len(metadata_results):,}"
)
log(
    f"Category thất bại   : {len(failed_categories):,}"
)
log(
    f"Sản phẩm cần tìm    : {total_wanted:,}"
)
log(
    f"Sản phẩm tìm thấy   : {total_found:,}"
)
log(
    f"Sản phẩm còn thiếu  : {total_missing:,}"
)
log(
    f"Sản phẩm đã ghi DB  : {total_written:,}"
)
log(
    f"Metadata đã quét    : {total_scanned:,} dòng"
)
log(
    f"Tổng thời gian      : {overall_elapsed / 60:.1f} phút"
)

if failed_categories:
    log("")
    log("Các category bị lỗi:")

    for category, error in failed_categories:
        log(f"  - {category}: {error}")

log("=========================================")
````

````python
FULFILLMENT = re.compile(
    r"^(good|great|excellent|perfect|nice|ok|okay|love it|five stars)[.! ]*$|"
    r"\b(as expected|fast shipping|arrived on time|would buy again)\b", re.I)
HISTORIES = WORK / "user_histories.prepared.jsonl"

with HISTORIES.open("w", encoding="utf-8") as out:
    for rank_no, user_id in db.execute("SELECT rank,user_id FROM selected_users ORDER BY rank"):
        reviews, seen = [], set()
        for r in db.execute("""SELECT review_id,category,parent_asin,asin,ts,rating,title,text,
            verified,helpful,product_title,product_main_category,product_category_path
            FROM reviews WHERE user_id=? ORDER BY ts,review_id""", (user_id,)):
            rid, cat, parent, asin, ts, rating, title, text, verified, helpful, pt, pm, pp = r
            key = (parent, asin, ts, rating, title, text)
            if key in seen or rating is None or FULFILLMENT.search((title+" "+text).strip()):
                continue
            seen.add(key)
            reviews.append({"review_id":rid,"category":cat,"parent_asin":parent,"asin":asin,
                "timestamp":ts,"rating":rating,"title":title,"text":text,
                "verified_purchase":bool(verified),"helpful_vote":helpful,
                "product_title":pt,"product_main_category":pm,
                "product_category_path":json.loads(pp or "[]")})
        if len(reviews) < 2: continue
        cut = max(1, min(int(len(reviews)*TRAIN_FRACTION), len(reviews)-1))
        out.write(json.dumps({"source":"amazon_reviews_2023","user_id":user_id,"rank":rank_no,
            "temporal_split":{"method":"per_user_temporal","train_fraction":TRAIN_FRACTION},
            "reviews":reviews[:cut],"validation_reviews":reviews[cut:]}, ensure_ascii=False)+"\n")
print("Prepared histories:", HISTORIES)
````

````python
HISTORIES = WORK / "user_histories.prepared.jsonl"
MODEL = "Qwen3-14B"
MAX_LLM_USERS = 3
MAX_REVIEWS_IN_PROMPT = 1

LLM_ENDPOINT = "http://203.113.152.4:7777/llm/v1/chat/completions"
LLM_AUTHORIZATION = "Basic dmlldHRlbF9haTpWYWlAMjAyNQ=="
LLM_TIMEOUT_SECONDS = 300

PERSONAS = WORK / "personas_1290.jsonl"
EVIDENCE = WORK / "evidence_profiles.jsonl"


def parse_llm_json(content: str) -> dict[str, Any]:
    content = content.strip()
    if content.startswith("```"):
        lines = content.splitlines()
        content = "\n".join(lines[1:-1]).strip()

    try:
        return json.loads(content)
    except json.JSONDecodeError:
        first_brace = content.find("{")
        last_brace = content.rfind("}")
        if first_brace < 0 or last_brace <= first_brace:
            raise
        return json.loads(content[first_brace:last_brace + 1])


def json_call(system: str, payload: dict[str, Any]) -> dict[str, Any]:
    request_body = {
        "model": MODEL,
        "messages": [
            {"role": "system", "content": system},
            {"role": "user", "content": json.dumps(payload, ensure_ascii=False)},
        ],
        "max_tokens": 5000,
        "temperature": 0.8,
        "top_p": 0.9,
        "top_k": 30,
        "presence_penalty": 1,
        "chat_template_kwargs": {"enable_thinking": False},
    }
    response = requests.post(
        LLM_ENDPOINT,
        headers={
            "Content-Type": "application/json",
            "Authorization": LLM_AUTHORIZATION,
        },
        json=request_body,
        timeout=LLM_TIMEOUT_SECONDS,
    )
    response.raise_for_status()
    response_data = response.json()
    content = response_data["choices"][0]["message"]["content"]
    return parse_llm_json(content)
def choose_reviews(reviews: list[dict[str, Any]]) -> list[dict[str, Any]]:
    def score(review: dict[str, Any]) -> int:
        text = review.get("text") or ""
        return (
            min(len(text), 500)
            + 50 * bool(re.search(r"\b(I|my|we|our)\b", text, re.I))
            + min(int(review.get("helpful_vote") or 0), 20) * 5
        )

    picked = sorted(reviews, key=score, reverse=True)[:MAX_REVIEWS_IN_PROMPT]
    return sorted(picked, key=lambda review: review["timestamp"])


def evidence_profile(user: dict[str, Any]) -> dict[str, Any]:
    rows = [
        {
            key: review.get(key)
            for key in (
                "review_id", "category", "rating", "title", "text",
                "product_title", "product_main_category", "product_category_path",
            )
        }
        for review in choose_reviews(user["reviews"])
    ]
    return json_call(
        "Build a compact evidence profile from Amazon reviews. Never infer sensitive status "
        "from products. Each claim must cite review_ids and short exact quotes. Return JSON.",
        {
            "user_id": user["user_id"],
            "reviews": rows,
            "output": {
                "evidence_items": [
                    {
                        "id": "e1",
                        "claim": "...",
                        "review_ids": ["..."],
                        "quotes": ["..."],
                        "confidence": 0.0,
                    }
                ],
                "summary": "...",
            },
        },
    )


with HISTORIES.open(encoding="utf-8") as history_file, EVIDENCE.open(
    "w", encoding="utf-8"
) as evidence_file:
    for user_index, line in enumerate(
        tqdm(history_file, desc="Creating evidence"), start=1
    ):
        if user_index > MAX_LLM_USERS:
            break
        user = json.loads(line)
        profile = evidence_profile(user)
        evidence_file.write(
            json.dumps(
                {"user_id": user["user_id"], "profile": profile},
                ensure_ascii=False,
            )
            + "\n"
        )

print("Evidence profiles:", EVIDENCE)
````

````python
DIMENSIONS_PER_CALL = 200

def infer_batch(
    user_id: str,
    profile: dict[str, Any],
    dims: list[dict[str, Any]],
) -> list[dict[str, Any]]:
    compact_dims = [
        {
            "id": dim["id"],
            "label": dim["label"],
            "description": dim["description"],
            "category": dim["category"],
            "allowed_values": dim["values"],
        }
        for dim in dims
    ]
    result = json_call(
        "Map grounded evidence to categorical persona dimensions. Omit unsupported dimensions. "
        "Use exactly one allowed value and cite construction review_ids. Return JSON.",
        {
            "user_id": user_id,
            "evidence_profile": profile,
            "dimensions": compact_dims,
            "output": {
                "attributes": [
                    {
                        "dimension_id": "...",
                        "value": "...",
                        "confidence": 0.0,
                        "evidence_review_ids": ["..."],
                        "reasoning": "...",
                    }
                ]
            },
        },
    )
    return result.get("attributes") or []


with EVIDENCE.open(encoding="utf-8") as evidence_file:
    evidence_by_user = {
        record["user_id"]: record["profile"]
        for line in evidence_file
        if line.strip()
        for record in [json.loads(line)]
    }

with HISTORIES.open(encoding="utf-8") as history_file, PERSONAS.open(
    "w", encoding="utf-8"
) as persona_file:
    for line in tqdm(history_file, desc="Creating personas"):
        user = json.loads(line)
        user_id = user["user_id"]
        profile = evidence_by_user.get(user_id)
        if profile is None:
            print(f"Skip {user_id}: missing evidence profile")
            continue

        valid_review_ids = {review["review_id"] for review in user["reviews"]}
        accepted: dict[str, dict[str, Any]] = {}

        for start in range(0, len(schema), DIMENSIONS_PER_CALL):
            dimensions = schema[start:start + DIMENSIONS_PER_CALL]
            for attr in infer_batch(user_id, profile, dimensions):
                dim = schema_by_id.get(attr.get("dimension_id"))
                evidence_ids = attr.get("evidence_review_ids") or []
                try:
                    confidence = float(attr.get("confidence"))
                except (TypeError, ValueError):
                    continue

                if (
                    dim
                    and attr.get("value") in dim["values"]
                    and evidence_ids
                    and set(evidence_ids).issubset(valid_review_ids)
                    and 0 <= confidence <= 1
                ):
                    accepted[dim["id"]] = attr

        fields = []
        for dim in schema:
            attr = accepted.get(dim["id"])
            fields.append(
                {
                    "field_id": dim["id"],
                    "value": attr.get("value") if attr else None,
                    "confidence": round(float(attr.get("confidence")), 3) if attr else 0.0,
                    "evidence_review_ids": attr.get("evidence_review_ids", []) if attr else [],
                    "reasoning": attr.get("reasoning", "") if attr else "",
                    "assignment_type": "summary_inference" if attr else "unsupported",
                }
            )

        assert len(fields) == 1290
        persona_file.write(
            json.dumps(
                {
                    "user_id": user_id,
                    "source": "amazon_reviews_2023",
                    "fields": fields,
                },
                ensure_ascii=False,
            )
            + "\n"
        )

print("Final personas:", PERSONAS)
````
